In this step, I focused on cleaning the dataset by checking and removing any missing values.
Real-world datasets, such as the Online Retail Dataset, often contain missing or incomplete entries — especially in columns like CustomerID and Description.

To handle this, I used the following commands:

df.isnull().sum() → to check how many missing (null) values exist in each column.

df.dropna() → to remove all the rows that contain missing values.

df.shape → to check the size of the dataset after cleaning, showing how many rows and columns remain.

After running this code, the dataset became cleaner and more reliable for further analysis and visualization. Removing missing data ensures better accuracy in the recommendation system and helps prevent errors in later steps.

```python
# Missing values check
print(df.isnull().sum())

# Missing values hatana (CustomerID aur Description me missing hoti hain)
df = df.dropna()

# Data ka size check karna
print("\nData shape after cleaning:", df.shape)
```

In this step, I created a bar chart to identify the top 10 best-selling products based on the total quantity sold.
Using the groupby() and nlargest() functions, I calculated which products had the highest sales.
The Seaborn barplot was then used to visualize these results clearly.

📝 Insight:
This chart helps to see which products are selling the most. Usually, items like stationery, decoration, or accessories appear at the top — focusing on these can help increase profit.

```python
top_products = df.groupby('Description')['Quantity'].sum().nlargest(10)

plt.figure(figsize=(10,5))
sns.barplot(x=top_products.values, y=top_products.index, palette='Blues')
plt.title("Top 10 Best-Selling Products")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product Description")
plt.show()
```

In this step, I created a pie chart to show the top 5 countries with the highest total sales quantity.
The data was grouped by the ‘Country’ column, and the top-performing countries were displayed as percentage shares of total sales.

📝 Insight:
This chart shows which countries contribute the most to overall sales. In this dataset, the United Kingdom is usually the largest market with the highest sales volume.

```python
country_sales = df.groupby('Country')['Quantity'].sum().nlargest(5)

plt.figure(figsize=(6,6))
country_sales.plot(kind='pie', autopct='%1.1f%%', colors=['#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ff9999'])
plt.title("Top 5 Countries by Sales Quantity")
plt.ylabel("")
plt.show()
```

In this step, I converted the InvoiceDate column into a proper date format and extracted the Month from it. Then, I grouped the data by month to calculate the total quantity sold.

A line chart was created to show the monthly sales trend.

📝 Insight:
This trend shows which months had higher sales. A visible spike in December indicates that sales increase during the holiday season when customer demand is high.

```python
# Convert 'InvoiceDate' to datetime safely
df.loc[:, 'InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Create a 'Month' column safely
df.loc[:, 'Month'] = df['InvoiceDate'].dt.to_period('M')

# Group by month and calculate total quantity sold
monthly_sales = df.groupby('Month')['Quantity'].sum()

# Plot the monthly sales trend
plt.figure(figsize=(10,5))
monthly_sales.plot(kind='line', marker='o', color='orange')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Quantity Sold")
plt.show()
```

In this step, I created a scatter plot to visualize the relationship between Unit Price and Quantity Sold.
A random sample of 500 records was used to make the chart clear and easy to read.

📝 Insight:
The plot shows an inverse relationship — expensive items are sold in smaller quantities, while cheaper products are sold in larger quantities.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='UnitPrice', y='Quantity', data=df.sample(500), color='purple')
plt.title("Unit Price vs Quantity Sold")
plt.xlabel("Price per Item")
plt.ylabel("Quantity Sold")
plt.show()
```

In this step, I created a histogram to show how product prices are distributed in the dataset.
The x-axis represents the unit price, and the y-axis shows the frequency of products in each price range.
Outliers above 50 were excluded to make the chart more readable.

📝 Insight:
Most products fall within the 0–20 price range, showing that the company mainly sells affordable, low-cost items.

```python
plt.figure(figsize=(8,5))
plt.hist(df['UnitPrice'], bins=50, color='green', edgecolor='black')
plt.xlim(0, 50)  # Outliers hata kar graph readable banate hain
plt.title("Distribution of Product Prices")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.show()
```
